```python
from typing import Self

mock = True

class Series:

    def __init__(self: Self, n: int, a: int) -> None:
        self.soma = self.partial_sum(n, a)

    def partial_sum(self: Self,  n: int, a: int) -> int:
        if a == 1:
            return n * (n + 1) // 2

        numerador = a - (n + 1) * pow(a, n + 1) + n * pow(a, n + 2)
        denominador = (1 - a) ** 2

        return numerador // denominador


if __name__ == "__main__":

    if mock:
        (n, a), content_output = prepare_data(exercise_id="01", sample_number=1, exercise_function_rules=exercise_01)

    else:
        n, a = map(int, input().split())

    series = Series(n, a)
    print(series.soma)
```
