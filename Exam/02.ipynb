## Problem Statement:
Exam session is a difficult time for everybody. Your friend (not you, of course) has problems with preparing to the next exam: he knows nothing and has only 
M minutes left before the exam. He has list of questions for the exam and for each question he has estimation of number of minutes he needs to study materials related to this question.

On the exam he will be given one question from this list. If he have been studied materials for the given question — he will pass the exam, otherwise he will fail.

Help your friend to maximise probability of passing the exam. Write a program that will calculate maximum number of questions your friend will manage to study materials for before the exam.

## Input:
First Line:
    1. Number of questions in the list (N: 0 < N ≤ 10^5)
    2. Number of minutes left before the exam (M: 0 < M ≤ 10^9)
Second Line:
    1. N integers separated by spaces: time estimations for each question (mi: 0 < mi ≤ 10^4)

## Output:
Single integer: maximum number of questions your friend can study before the exam.


```python
from DataHandler import prepare_data

def exercise_rules(raw_input: list[str]) -> list[int]:
    N, M = raw_input[0].strip().split()
    mi = [int(x) for x in raw_input[1].strip().split()]
    return int(N), int(M), mi

(N, M, mi), content_output = prepare_data(
    exercise_id="02",
    sample_number=1,
    exercise_function_rules=exercise_rules,
)

print(N, M, mi)
```

```python
# N: Number of questions
# M: Minutes available
# mi: List of time required for each question

from typing import Self


class TightSchedule:

    def __init__(self: Self, n: int, m: int, mi: list[int]) -> Self:
        self.n = n
        self.m = m
        self.mi = mi
        self.max_number_questions = self.solve()

    def solve(self: Self) -> int:

        count = 0
        mi_sorted = sorted(self.mi)

        for time_per_question in mi_sorted:
            if self.m - time_per_question >= 0:
                self.m -= time_per_question
                count += 1
            else:
                break

        return count

if __name__ == "__main__":

    MOCK = True

    if MOCK:
        EXERCISE_ID = "02"
        SAMPLE_NUMBER = 1
        FUNCTION_RULES = exercise_rules
        PRINT_DATA_BEFORE_RETURN = False
        (N, M, mi), content_output = prepare_data(
            exercise_id=EXERCISE_ID, 
            sample_number=SAMPLE_NUMBER, 
            exercise_function_rules=FUNCTION_RULES,
            print_data_before_return=PRINT_DATA_BEFORE_RETURN,
        )

        exercise = TightSchedule(N, M, mi)
        print(exercise.max_number_questions)

    else:
        N, M = map(int, input().split())
        mi = list(map(int, input().split()))
        exercise = TightSchedule(N, M, mi)
        print(exercise.max_number_questions)
        pass

```

```python
weights = [10, 20, 30]
values = [200, 100, 120]

capacity = 50

def knapsack(weights, values, capacity):
    n = len(weights)
    dp = [0] * (capacity + 1)

    for item in range(n):
        for weight in range(capacity, weights[item] - 1, -1):
            dp[weight] = max(dp[weight], dp[weight - weights[item]] + values[item])

    return dp, dp[capacity]

dp, dp_capacity = knapsack(weights, values, capacity)
print(dp)
print(dp_capacity)
```
