# Exercise 07 (G):

You’re still very interested in different range queries and keep inventing new ones. Your new obsession is REQ — Range Equal Query. REQ(l, r, k) — is number of times value 
k occurs in range [l,r): 

`REQ(l,r,k)=∣{i:a[i]=k}∩[l;r)∣`

You have initial array a of 0 < N ≤ 100000 elements and 0 < M ≤ 100000 queries of two types:

`? l r k` — REQ(l,r,k)

or

`+ i delta` — inc(i,delta): increase a[i] value by delta.

It is guaranteed that elements of array never exceed K : 0 ≤ a_i ≤ K.

## Input format:

First line contains 3 integer numbers: 
0 < N, M ≤ 100000 ; 0 < K ≤ 100 — number of items in array, number of queries and limitation on values in array.

Next line contains N integer numbers — values of initial array a: 0 ≤ a_i ≤ K.

Each of next M lines contains character followed by 3 or 4 integer numbers — request of one of two types:

? l r k — REQ(l,r,k): 0 ≤ l < r ≤ N; 0 ≤ k ≤ K

or

+ i delta — inc(i,delta): increase a[i] value by delta. 0 ≤ i < N; −K ≤ k ≤ K

## Output format:

For each "?" query provide answer for REQ request.

```python
from collections import Counter
from typing import Self, List, Tuple


class RangeEqualQuery:

    def __init__(self: Self, a: List[int], queries: List[Tuple[str, Tuple[int, int, int]]]) -> None:
        self.a = a
        self.solve(queries)

    def inc(self: Self, i: int, delta: int) -> None:
        self.a[i] += delta

    def req(self: Self, l: int, r: int, k: int) -> int:
        a_filtered = self.a[l:r]
        count = Counter(a_filtered)
        return count[k]
    
    def solve(self: Self, queries: List[Tuple[str, Tuple[int, int, int]]]) -> None:
        for query in queries:
            query_type, params = query

            if query_type == "?":
                l, r, k = params
                result = self.req(l, r, k)
                print(result)

            elif query_type == "+":
                i, delta = params
                self.inc(i, delta)

    def __repr__(self):
        return f"A = {self.a}"
    
A = RangeEqualQuery(A0)
print(A)


for query in queries:
    query_type, params = query

    if query_type == "?":
        l, r, k = params
        result = A.req(l, r, k)
        print(result)
        # print(f"REQ({l},{r},{k}) = {result}")
    elif query_type == "+":
        i, delta = params
        A.inc(i, delta)
        # print(f"inc({i},{delta}) -> A = {A}")
```

```python
from collections import Counter
from typing import Self, List, Tuple


class FenwickTree:
    def __init__(self: Self, n: int) -> None:
        self.n = n
        self.tree = [0] * (n + 1)  # 1-based indexing

    def add(self: Self, index: int, delta: int) -> None:
        # index must be 1-based
        while index <= self.n:
            self.tree[index] += delta
            index += index & -index

    def prefix_sum(self: Self, index: int) -> int:
        # sum from 1 to index
        result = 0
        while index > 0:
            result += self.tree[index]
            index -= index & -index
        return result

    def range_sum(self, left: int, right: int) -> int:
        # sum from left to right (inclusive)
        return self.prefix_sum(right) - self.prefix_sum(left - 1)


a = [3, 1, 4, 1, 5]

fw = FenwickTree(len(a))

# build the tree
for i, value in enumerate(a):
    fw.add(i + 1, value)

print(fw.prefix_sum(3))    # sum of a[0..2] → 3 + 1 + 4 = 8
print(fw.range_sum(2, 4))  # sum of a[1..3] → 1 + 4 + 1 = 6

# update a[2] += 10
fw.add(3, 10)

print(fw.range_sum(2, 4))  # now 1 + 14 + 1 = 16
```
