## Exercise 08 (H):

You decided to visit MIPT. You successfully arrived in Moscow and came to Savelovskiy railway station. Now you are going to buy a ticket, but there are a lot of people who wants the same. They stand in a single huge queue.

Tickets are sold at K ticket windows, but the queue is common for all of these windows. When any window becomes free, first person from the queue comes to this window, then he buys a ticket and leaves.

You have estimated amount of time (in seconds) each person will spend to buy a ticket (amount of time he will occupy ticket window). When you came to station, all ticket windows were free (this mens, first K persons from the queue will go to windows immediately).

You want to calculate number of seconds that will pass before all the persons in the queue will obtain their tickets.

## Input format:

First line contains two integer numbers: 0 < N ≤ 10^5 ,0 < K ≤ 10^4 — number of persons in the queue and number of ticket windows in the station.
Next line contains N integer numbers divided by space character: 
0 < t_i ≤10^5 — number of seconds i-th person will occupy ticket window.

## Output format:

Print single integer number: number of seconds that will pass before all persons in the queue obtain their tickets.

```python
from DataHandler import prepare_data
def exercise_rules(raw_input: list[str]) -> list[int]:
    N, K = map(int, raw_input[0].strip().split())
    queue = [int(x.strip()) for x in raw_input[1].split()]
    return N, K, queue

# N: Number of people in the queue
# K: Number of ticket windows in the station
# t_i: Time taken to serve the i-th person
from typing import Self, List

class TicketQueue:

    def __init__(self: Self, n_windows: int) -> None:
        self.n_windows = n_windows
        self.windows: List[int] = [0] * n_windows
        self.solve(queue)

    def assign_person(self: Self, service_time: int) -> None:
        min_window_index = self.windows.index(min(self.windows))
        self.windows[min_window_index] += service_time

    def total_time(self: Self) -> int:
        return max(self.windows)

    def solve(self: Self, queue: List[int]) -> int:
        for t in queue:
            self.assign_person(t)
        return self.total_time()

EXERCISE_ID = "08"
SAMPLE_NUMBER = 1
FUNCTION_RULES = exercise_rules
PRINT_DATA_BEFORE_RETURN = True

(N, K, queue), content_output = prepare_data(
    exercise_id=EXERCISE_ID, 
    sample_number=SAMPLE_NUMBER, 
    exercise_function_rules=FUNCTION_RULES,
    print_data_before_return=PRINT_DATA_BEFORE_RETURN,
)

TicketQueue_instance = TicketQueue(K)
print(TicketQueue_instance.total_time())

windows: list[int] = [0] * K  # Time when each window will be free
for service_time in queue:
    # Find the window that will be free the earliest
    min_window_index = windows.index(min(windows))
    # Assign the person to that window
    windows[min_window_index] += service_time
print(max(windows))  # The total time taken is the time when the last window is free
```

```python
from typing import Self, List
from heapq import heappop, heappush, heapify

class TicketQueue:

    def __init__(self: Self, n_windows: int, queue: List[int]) -> None:
        self.n_windows = n_windows
        self.queue = queue
        self.solve()

    def solve(self: Self) -> None:

        heap = [0] * self.n_windows
        heapify(heap)

        for service_time in self.queue:
            earliest_free_time = heappop(heap)
            heappush(heap, earliest_free_time + service_time)
        
        print(max(heap))


if __name__ == "__main__":

    MOCK = True

    if MOCK:
        EXERCISE_ID = "08"
        SAMPLE_NUMBER = 3
        FUNCTION_RULES = exercise_rules

        (N, K, queue), content_output = prepare_data(
            exercise_id=EXERCISE_ID, 
            sample_number=SAMPLE_NUMBER, 
            exercise_function_rules=FUNCTION_RULES,
        )

        TicketQueue_instance = TicketQueue(K, queue)
        pass

    else:
        N, K = map(int, input().strip().split())
        queue = [int(x.strip()) for x in input().strip().split()]
        TicketQueue_instance = TicketQueue(K, queue)
```
