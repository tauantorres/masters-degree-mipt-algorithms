# Exercise 09 (I):

You are researching different DNAs. DNA is a sequence of nucleotides. There are 4 types of nucleotides in DNA: cytosine [C], guanine [G], adenine [A] or thymine [T].

During your previous experiment you found out that you should not consider DNAs which contain at least 3 Adenine nucleotides [A] in a row.

Now you want to calculate amount of different DNAs of N nucleotides which you still need to consider. (Don’t think about complementary sequence, just ignore complementarity rules).

## Input format:
Input file contains single integer number 0 < N ≤ 30.

## Output format:
Print single integer number: number of DNAs consisting of N nucleotides which doesn’t have 3 As in a row.

```python
from DataHandler import prepare_data
def exercise_rules(raw_input: list[str]) -> list[int]:
    N = int(raw_input[0].strip())
    return N

# N: Length of DNA sequence;
import pandas as pd

EXERCISE_ID = "09"
SAMPLE_NUMBER = 3
FUNCTION_RULES = exercise_rules
PRINT_DATA_BEFORE_RETURN = False

(N), content_output = prepare_data(
    exercise_id=EXERCISE_ID, 
    sample_number=SAMPLE_NUMBER, 
    exercise_function_rules=FUNCTION_RULES,
    print_data_before_return=PRINT_DATA_BEFORE_RETURN,
)

N = 10

print(f"Length of DNA sequence: {N}")

dp = [[0] * 3 for _ in range(N + 1)]
dp[0][0] = 1

for i in range(0, N):
    dp[i + 1][0] = 3 * (dp[i][0] + dp[i][1] + dp[i][2])
    dp[i + 1][1] = dp[i][0]
    dp[i + 1][2] = dp[i][1]

df = pd.DataFrame(dp)
print(f"Total number of valid DNA sequences: {sum(dp[-1])}")
df.head(N + 1)
```
