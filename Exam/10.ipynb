Unfortunately, your table is not vert big: it may contain only K books simultaneously. If you have K books on the table, and you need one which is not on the table — you should return one book from table to bookcase and replace it with needed one.

You want to concentrate on study, so you decided to minimize number of times you will take book from bookcase. You have enumerated all books with numbers [1; N] and prepared a list of 0 < M ≤ 10^5 books in order you will use them (one book can be present more than one time in this list). Calculate minimum number of times you should take book from bookcase according to given list.

## Input format:

First line contains 3 integer numbers: 0 < N, K , M ≤ 10^5 — number of books in the bookcase, maximum number of books on the table, number of items in the usage list.
Second line contains N integer numbers ∈[1,N] — ids of books in order you will use them.

## Output format:

Print single integer number — minimum number of times you will need to take a book from bookcase.

```python
def update_status(books_id: dict[int, bool], book_id: int, status: bool) -> None:
    books_id[book_id] = status

def get_index_of_next_usage(usage_list: list[int], book_id: int, start_index: int) -> int:
    try:
        index = usage_list[start_index:].index(book_id)
        return index + start_index
    except ValueError:
        return -1

def get_id_of_books_on_table(books_id: dict[int, bool]) -> list[int]:
    return [book_id for book_id, is_on_table in books_id.items() if is_on_table]

def get_furthest_to_use_book(usage_list: list[int], books_on_table: list[int], start_index: int) -> int:
    next_usages = []
    for book_id in books_on_table:
        next_usage_index = get_index_of_next_usage(usage_list, book_id, start_index)
        next_usages.append((book_id, next_usage_index))
    next_usages.sort(key=lambda x: x[1], reverse=True)
    return next_usages[-1][0] if next_usages[-1][1] == -1 else next_usages[0][0]

books_id = {id: False for id in range(1, N + 1)}
print(books_id)
print(f"Usage list: {usage_list}")

count = 0

for i in range(1, M + 1):

    current_book = usage_list[i - 1]

    if i - 1 < K:  # Table not full yet:

        if not books_id[current_book]:
            update_status(books_id, current_book, True)
            count += 1
            print(f"Placing book {current_book} on the table. Total placements: {count}")
        else:
            print(f"Book {current_book} is already on the table.")
    
    else:  # Table full:
        
        if not books_id[current_book]:  # Book not on table
            books_on_table = get_id_of_books_on_table(books_id)
            furthest_book_to_use = get_furthest_to_use_book(usage_list, books_on_table, i)
            update_status(books_id, furthest_book_to_use, False)
            update_status(books_id, current_book, True)
            count += 1
            print(f"Replacing book {furthest_book_to_use} with book {current_book}. Total placements: {count}")
        
        else:
            print(f"Book {current_book} is already on the table.")
            continue
    

print(f"Total placements on the table: {count}")
```
