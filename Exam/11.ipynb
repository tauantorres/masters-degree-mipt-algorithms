# Exercise 11 (J):

One famous Korean company, which is famous by it’s smartphones and skyscrapers (e.g. Burj Khalifa) needs your help.

Company wants to build two skyscrapers of identical height. Skyscrapers will have main pillar in the middle. This pillar should go vertically through each level (floor) of each skyscraper. Company already has materials for these pillars. These materials are N undividable parts of the pillar, i-th part has a height of h_i levels (floors). Company wants the total height (number of levels) of pillars for both skyscrapers to be identical and as big as possible.

Help company to choose parts of pillars which will allow both pillars to have maximum identical height.

## Input format:

First line of input file contains single integer number 0 < N ≤ 50 — number of parts of pillars.

Second line contains N integer numbers divided by space character: 0 < h_i ≤7 — height of i-th part of the pillar.

It is guaranteed, that two skyscrapers of positive identical height can always be built.

## Output format:

On the first line print desired maximum height of both pillars.

On the second line print number of parts to be used for the first skyscraper.

On the third line print indices of parts to be used for the first skyscraper.

On the fourth line print number of parts to be used for the second skyscraper.

On the fifth line print indices of parts to be used for the second skyscraper.

If there are several ways to provide desired height — you can print any one. You can print indices of parts in any order.

```python
from sys import maxsize

OFFSET = 350
INF = (-1) * maxsize

# dp[i][d] = max total height using first i blocks with diff = d - OFFSET
dp = [dict() for _ in range(N + 1)]
dp[0][OFFSET] = 0

MAX_HEIGHT = sum(pilars_height) // 2

# parent[i][d] = (previous_difference, choice)
parent = [dict() for _ in range(N + 1)]

for i in range(N):
    height = pilars_height[i]

    for diff, value in dp[i].items():

        if diff not in dp[i + 1] or dp[i + 1][diff] < value:
            dp[i + 1][diff] = value
            parent[i + 1][diff] = (diff, 0)

        # Add to first pilar
        new_diff = diff + height
        if 0 <= new_diff <= 2 * OFFSET:
            if new_diff not in dp[i + 1] or dp[i + 1][new_diff] < value + height:
                dp[i + 1][new_diff] = value + height
                parent[i + 1][new_diff] = (diff, 1)

        # Add to second pilar
        new_diff = diff - height
        if 0 <= new_diff <= 2 * OFFSET:
            if new_diff not in dp[i + 1] or dp[i + 1][new_diff] < value + height:
                dp[i + 1][new_diff] = value + height
                parent[i + 1][new_diff] = (diff, -1)

total_used = dp[N][OFFSET]
H = total_used // 2

A, B = [], []
diff = OFFSET
for i in range(N, 0, -1):
    prev_diff, choice = parent[i][diff]
    if choice == 1:
        A.append(i-1)
    elif choice == -1:
        B.append(i-1)
    diff = prev_diff

print(pilars_height)
print(H)
print(len(A))
print(*A)
print(len(B))
print(*B)

```
