## Working with the Data

Now you can work with your scraped data as usual:

```python
# Example: Process the input data
def prepare_data(
    exercise_number: str = "08",
    sample_number: int = 1,
):
    content_input, content_output = get_data(
        exercise_number=exercise_number,
        sample_number=sample_number,
    )
    
    # Parse the input
    lines = [line.strip() for line in content_input]
    
    # First line: N, M, K
    n, m, k = map(int, lines[0].split())
    print(f"N={n}, M={m}, K={k}")
    
    # Second line: initial array
    initial_array = list(map(int, lines[1].split()))
    print(f"Initial array: {initial_array}")
    
    # Parse queries
    queries = []
    for i in range(2, len(lines)):
        if lines[i]:
            parts = lines[i].split()
            queries.append(parts)
    
    print(f"Queries: {queries}")
    
    # Parse expected output
    expected = [line.strip() for line in content_output if line.strip()]
    print(f"Expected output: {expected}")
    
    return n, m, k, initial_array, queries, expected

# Test with our scraped data
n, m, k, arr, queries, expected = prepare_data("08", 1)
```
