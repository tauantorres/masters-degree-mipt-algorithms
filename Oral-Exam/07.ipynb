# `07 — Dynamic Programming (DP)`

We study:
- **Basic principles of DP**

  - naive recursion (**O(2^N)**)
  - top-down memoization (**O(N)**)
  - bottom-up tabulation (**O(N)**)
- **Longest Increasing Subsequence (LIS)**:
  - DP **O(N^2)**
  - optimized **O(N log N)** (binary search)

Goal:
- Know the **DP recipe**
- Know the **Big-O** of each approach

## `1. Basic principles of DP`

### `1.1 DP works when we have:`

#### **(1) Overlapping subproblems**
The same smaller problems appear many times.
Example: Fibonacci recursion recomputes fib(30) again and again.

#### **(2) Optimal substructure**
An optimal solution can be built from optimal solutions of subproblems.
Usually proven by contradiction.

### `1.2 DP recipe (how to build a DP solution)`

1) Define subproblems (parameterize them), e.g. **d[i]**
2) Define base cases, e.g. **d[0]**, **d[1]**
3) Define recurrence (transition), e.g. **d[i] = F( d[i-1] , d[i-2], ... )**
4) Compute answers using:
   - **Top-down** (memoization) or
   - **Bottom-up** (tabulation)

## `2. Fibonacci — 3 approaches`

### `2.0 Fibonacci complexity summary`

| Approach | Idea | Time | Memory |
|---|---|---:|---:|
| **Naive recursion** | recompute everything | **O(2^N)**| O(N) stack |
| **Top-down memo** | cache results | **O(N)**| O(N) |
| **Bottom-up tabulation** | fill dp table | **O(N)**| O(N) (or O(1) optimized) |


```python
# Naive Recursion for Fibonacci Numbers:

MOD: int = 10**9

def fib_naive(n: int, *, verbose: bool = False, depth: int = 0) -> int:
    if verbose:
        print("  " * depth + f"fib({n})")
    if n < 2:
        return n
    return (fib_naive(n - 1, verbose=verbose, depth=depth + 1) +
            fib_naive(n - 2, verbose=verbose, depth=depth + 1)) % MOD


print("-" * 55)
print(f"{'Fibonacci naive — Example trace':^55}")
print("-" * 55)
print("fib(6) =", fib_naive(6, verbose=True))
```

```python
#  Top-down Memoization for Fibonacci Numbers:
from typing import Dict


def fib_top_down(n: int, *, memo: Dict[int, int], verbose: bool = False) -> int:
    if n in memo:
        return memo[n]
    if verbose:
        print(f"compute fib({n})")
    memo[n] = (fib_top_down(n - 1, memo=memo, verbose=verbose) +
               fib_top_down(n - 2, memo=memo, verbose=verbose)) % MOD
    return memo[n]


print("-" * 55)
print(f"{'Fibonacci top-down — memo growth':^55}")
print("-" * 55)

memo = {0: 0, 1: 1}
for n in range(2, 11):
    val = fib_top_down(n, memo=memo, verbose=True)
    print(f"  fib({n}) = {val}, memo_size={len(memo)}")
```

**Note takeaway**: Each fib(k) is computed once, so total work is O(N).


```python
# Buttom-up Tabulation for Fibonacci Numbers:
from typing import List


def fib_bottom_up(n: int, *, verbose: bool = False) -> int:
    tab: List[int] = [0] * max(n + 1, 2)
    tab[0] = 0
    tab[1] = 1

    if verbose:
        print(f"tab[0]={tab[0]}, tab[1]={tab[1]}")

    for i in range(2, n + 1):
        tab[i] = (tab[i - 1] + tab[i - 2]) % MOD
        if verbose and i <= 12:
            print(f"tab[{i}] = tab[{i-1}] + tab[{i-2}] = {tab[i]}")

    return tab[n]


print("-" * 55)
print(f"{'Fibonacci bottom-up — Example trace':^55}")
print("-" * 55)
print("fib(10) =", fib_bottom_up(10, verbose=True))
```

```python
# Optional O(1) memory optimization for Fibonacci Numbers:

def fib_bottom_up_optimized(n: int) -> int:
    if n < 2:
        return n
    a: int = 0
    b: int = 1
    for _ in range(2, n + 1):
        a, b = b, (a + b) % MOD
    return b

print("fib(10) optimized =", fib_bottom_up_optimized(10))
```

## `3. Longest Increasing Subsequence (LIS)`

### `3.1 LIS statement`

Given a sequence $x[0..N-1]$, find the maximum length $K$ such that there exist indices:

$$0 \leq i_0 < i_1 < \ldots < i_{K-1} < N$$

with strictly increasing values:

$$x[i_0] < x[i_1] < \ldots < x[i_{K-1}]$$


## `4. LIS — DP Solution`

### `4.1 DP idea (O(N^2))`

**Define:**  
$$d[i] = \text{length of LIS that ends exactly at } x[i]$$

**Base:**  
$$d[0] = 1$$

**Transition:**  
$$d[i] = 1 + \max_{j < i,\, x[j] < x[i]} d[j]$$

If no such $j$ exists $\Rightarrow d[i] = 1$

**Answer:**  
$$\max_{i} d[i]$$

```python
from typing import List, Tuple


def lis_n2_trace(x: List[int], *, verbose: bool = True) -> Tuple[int, List[int]]:
    n: int = len(x)
    if n == 0:
        return 0, []

    d: List[int] = [1] * n  # d[i] >= 1 always

    if verbose:
        print("x:", x)
        print("d:", d)
        print("-" * 55)

    for i in range(n):
        best: int = 0
        for j in range(i):
            if x[j] < x[i] and d[j] > best:
                best = d[j]
        d[i] = best + 1

        if verbose:
            print(f"i={i:>2}, x[i]={x[i]:>3} -> d={d}")

    return max(d), d


print("-" * 55)
print(f"{'LIS O(N^2) — Example 01':^55}")
print("-" * 55)

length, d = lis_n2_trace([3, 5, 10, 1, 6, 8, 9, 8], verbose=True)
print("-" * 55)
print("LIS length =", length)
print("-" * 55)
```

## `5. LIS in O(N log N) (tails array + binary search)`

### `5.1 O(N log N) idea`

We maintain an array d where:

**d[l] = minimum possible last value**
among all increasing subsequences of length l
seen so far.

We keep:
- d[0] = -∞
- d[1..N] = +∞

For each x[i], find:
l* = min { l : d[l] >= x[i] }   (binary search)
then set:
d[l*] = x[i]

Because d is always increasing (monotone),
binary search works → O(log N) per element.
Total: O(N log N).

```python
from sys import maxsize
from bisect import bisect_left
from typing import List, Tuple


INF: int = maxsize


def lis_nlogn_trace(x: List[int], *, verbose: bool = True) -> Tuple[int, List[int]]:

    n: int = len(x)
    d: List[int] = [INF] * (n + 1)
    d[0] = -INF  # Smaller than any x[i]

    if verbose:
        print("x:", x)
        print("d (start):", d)
        print("-" * 55)

    for i, val in enumerate(x):
        # Find first l such that d[l] >= val
        l_star: int = bisect_left(d, val)
        d[l_star] = val

        if verbose:
            # Show only "meaningful prefix" of d
            prefix = []
            for t in d:
                if t >= INF // 2:
                    prefix.append("∞")
                else:
                    prefix.append(str(t))
            print(f"i={i:>2}, x[i]={val:>3} -> l*={l_star}, d={prefix}")

    # Answer = max l such that d[l] < INF
    ans: int = 0
    for l in range(n, -1, -1):
        if d[l] < INF:
            ans = l
            break

    return ans, d


print("-" * 55)
print(f"{'LIS O(N log N) — Example 01':^55}")
print("-" * 55)


length, d = lis_nlogn_trace([3, 5, 10, 1, 6, 8, 9, 8], verbose=True)
print("-" * 55)
print("LIS length =", length)
print("-" * 55)
```

## `6. Summary`

- **LIS O(N^2)**:
  - Easy DP definition (**d[i] ends at i**)
  - Works well for N up to maybe ~5000 in Python

- **LIS O(N log N)**:
  - Uses a clever “tails” array (**minimum last value**)
  - Uses binary search because d is monotone
  - Much faster for large N

Both return the **length**; reconstructing the actual subsequence needs extra parent pointers (not required here unless your exam asks).
