```python
from Tools import prepare_raw_data, get_data

prepare_raw_data(
    exercise_id="A",
    sample_number=3,
)

content_input, content_output = get_data()

def prepare_data():

    content_input, content_output = get_data()

    # Exercise Rules:
    # N: Number of Cities
    # M: Number of Roads
    # s: Start City
    # h: Home City
    # roads (u, v, w): Road from city u to city v with length w

    N, M = tuple(map(int, content_input[0].split()))
    s, h = tuple(map(int, content_input[1].split()))
    roads = [list(map(int, line.split())) for line in content_input[2: M+2]]
    
    roads_dist: dict[dict[int]] = {}

    for u, v, w in roads:

        if u not in roads_dist:
            roads_dist[u] = {}

        if v not in roads_dist:
            roads_dist[v] = {}

        roads_dist[u][v] = w
        roads_dist[v][u] = w

    return N, M, s, h, roads_dist
```
