```python
from Tools import prepare_raw_data, get_data

prepare_raw_data(
    exercise_id="B",
    sample_number=4,
)


def prepare_data():

    content_input, content_output = get_data()

    output = int(content_output[0])

    # Exercise Rules:
    # N: Number of Cities in the Schedule
    # s: id of Home City
    # f: id of Destination City
    # K: Maximum Number of Flights Allowed (including layovers)
    # aij: Adjacency Matrix representing direct flight costs between cities
    N, s, f, K = map(int, content_input[0].split())
    aij = [list(map(int, line.split())) for line in content_input[1 : N+1]]

    return N, s, f, K, aij, output
```

```python
import sys


INF = sys.maxsize


def generate_dp_matrix(N: int, K: list[list[int]]) -> list[list[int]]:
    return [[INF] * N for _ in range(K + 1)]

def student_id(N: int, s: int, f: int, K: int, cost: list[list[int]]) -> None:
    
    dp = generate_dp_matrix(N, K)

    dp[0][s] = 0

    for k in range(K):
        for u in range(N):

            if dp[k][u] == INF:
                continue
            
            for v in range(N):
                if cost[u][v] != -1:
                    dp[k + 1][v] = min(dp[k + 1][v], dp[k][u] + cost[u][v])

    answer = min(dp[k][f] for k in range(K + 1))

    if answer == INF:
        print(-1)
    else:
        print(answer)


if __name__ == '__main__':

    # N, s, f, K, cost, output = prepare_data()
    N, s, f, K = map(int, input().split())
    cost = [list(map(int, input().split())) for _ in range(N)]
    student_id( N, s, f, K, cost)
```
