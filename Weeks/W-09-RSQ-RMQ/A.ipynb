```python
from sys import maxsize
from typing import Self

INF = maxsize

class SegmentTree:

    def __init__(self: Self, A: list[int]) -> None:
        self.N = len(A)
        self.size = 1

        while self.size < self.N:
            self.size *= 2

        self.tree = [INF] * (2 * self.size)

        # Build Leaves:
        for i in range(self.N):
            self.tree[self.size + i] = A[i]

        # Build Internal Nodes:
        for i in range(self.size - 1, 0, -1):
            self.refresh_tree(i)

    def refresh_tree(self: Self, index: int) -> None:
        self.tree[index] = min(self.tree[2 * index], self.tree[2 * index + 1])

    def update(self: Self, index: int, value: int) -> None:
        index += self.size
        self.tree[index] = value

        while index > 1:
            index //= 2
            self.refresh_tree(index)

    def query(self: Self, left: int, right: int) -> int:
        
        result = INF
        left += self.size
        right += self.size

        while left < right:
            # if l is odd: "(left & 1) == (left % 2 == 1)" <- This two means the same.
            if left & 1:
                result = min(result, self.tree[left])
                left += 1

            if right & 1:
                right -= 1
                result = min(result, self.tree[right])

            left //= 2
            right //= 2

        return result


if __name__ == "__main__":

    N, M, A, Queries, raw_output = prepare_data()
    print("Expected Output:", raw_output)
    # N, M = map(int, input().split())
    # A = list(map(int, input().split()))
    # Queries = [input().split() for _ in range(M)]

    segment_tree = SegmentTree(A)

    for query in Queries:

        type_query= query[0]

        if type_query == "+":
            index, value = int(query[1]), int(query[2])
            current_value = segment_tree.tree[segment_tree.size + index]
            segment_tree.update(index, value)

        elif type_query == "?":
            left, right = int(query[1]), int(query[2])
            result = segment_tree.query(left, right)
            print(result)
```

```python
from math import pow
from time import time
from typing import Callable, Any


def timer(function: Callable) -> Callable:
    def wrapper(*args: Any, **kwargs: Any) -> Any:
        start = time()
        result = function(*args, **kwargs)
        end = time()
        print(f"Time taken: {end - start:.10f} seconds")
        return result
    return wrapper

@timer
def power(x: int, n: int) -> int:
    return int(pow(x, n))

@timer
def power_iterative(x: int, n: int) -> int:
    result = 1
    for _ in range(n):
        result *= x
    return result

@timer
def power_recursive(x: int, n: int) -> int:
    if n == 0:
        return 1
    elif n % 2 == 0:
        half_power = power_recursive(x, n // 2)
        return half_power * half_power
    else:
        return x * power_recursive(x, n - 1)


@timer
def power_fast_exponentiation(x: int, n: int) -> int:
    result = 1
    while n > 0:
        if n & 1:
            result *= x
        x *= x
        n >>= 1
    return result


z = power(2, 1000)
print('-' * 100)

z = power_iterative(2, 1000)
print('-' * 100)

z = power_fast_exponentiation(2, 1000)
print('-' * 100)

```

```python
def exponential(x: int) -> int:
    return x * x

x = 3
n = 3
expected = x ** n

base = x  # Store original base
result = 1

print(f"Computing {x}^{n} using fast exponentiation:")

while n > 0:
    if n & 1:  # If n is odd
        result *= base
        print(f"n is odd, multiply result by {base}: result = {result}")
    
    base = exponential(base)  # Square the base
    n >>= 1  # Divide n by 2
    
    if n > 0:
        print(f"Square base to {base}, n = {n}")

print(f"Final Result: {result}")
print(f"Expected Result: {expected}")
```
