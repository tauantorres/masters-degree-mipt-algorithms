```python
from typing import Self

N, M, A, Queries, output = prepare_data()
print(output)

class FenwickTree:

    def __init__(self: Self, n: int) -> None:
        self.size = n
        self.tree = [0] * (n + 1)

    def add(self: Self, index: int, value: int) -> None:
        
        index += 1

        while index <= self.size:
            self.tree[index] += value
            index += index & -index

    def sum(self: Self, index: int) -> int:

        index += 1
        result = 0

        while index > 0:
            result += self.tree[index]
            index -= index & -index
        
        return result

    def range_sum(self: Self, left: int, right: int) -> int:
        return self.sum(right - 1) - (self.sum(left - 1) if left > 0 else 0)

class RZQ:

    def __init__(self: Self, a: list[int]) -> None:
        self.a = a
        self.n = len(a)
        self.tree = FenwickTree(self.n)

        for i in range(self.n):
            if a[i] == 0:
                self.tree.add(i, 1)

    def rzq(self: Self, left: int, right: int) -> int:
        return self.tree.range_sum(left, right)
    
    def update(self: Self, index: int, value: int) -> None:
        was_zero = (self.a[index] == 0)

        self.a[index] += value

        is_zero = (self.a[index] == 0)

        if was_zero and not is_zero:
            self.tree.add(index, -1)
        elif not was_zero and is_zero:
            self.tree.add(index, 1)

if __name__ == "__main__":

    # N, M = map(int, input().split())
    # A = list(map(int, input().split()))
    # Queries = [input().split() for _ in range(M)]
    N, M, A, Queries, output = prepare_data()

    A = RZQ(A)
    for query in Queries:

        process = query[0]
        if process == "?":
            l, r = int(query[1]), int(query[2])
            zeros_found = A.rzq(l, r)
            print(zeros_found)

        elif process == "+":
            index, value = int(query[1]), int(query[2])
            A.update(index, value)
```

```python
from typing import Self

N, M, A, Queries, output = prepare_data()
print(N, M, A, Queries, output)

class RZQ:

    def __init__(self: Self, a: list[int]) -> None:
        self.a = a

    def rzq(self: Self, left: int, right: int) -> int:
        zeros = 0
        for value in self.a[left:right]:
            if value == 0:
                zeros += 1
        return zeros
    
    def update(self: Self, index: int, value: int) -> None:
        self.a[index] = self.a[index] + value

A = RZQ(A)
print("Initial array state:", A.a, '\n')
for query in Queries:

    process = query[0]
    print("Processing query:", query)
    print(" - Type:", "(?) Range Zero Query" if process == "?" else "(+) Update Query")
    print(" - [BEFORE] Current array state:", A.a)

    if process == "?":
        l, r = int(query[1]), int(query[2])
        zeros_found = A.rzq(l, r)
        print(" - Zeros found in range [{}, {}): {}".format(l, r, zeros_found))
        
    elif process == "+":
        index, value = int(query[1]), int(query[2])
        A.update(index, value)

    print(" - [AFTER] Current array state:", A.a, '\n')
```

```python
from sys import maxsize
from typing import Self

class RSQPrefixSum:
    """
    Complexity:
        - Preprocessing Time: O(N)
        - Query Time: O(1)
        - Update: O(N) (not efficient for frequent updates)
    """
    
    INF = maxsize

    def __init__(self: Self, data: list[int]) -> None:
        self.sum = []
        for value in data:
            self.sum.append(self.sum[-1] + value if self.sum else value)
    
    def rsq(self: Self, left: int, right: int) -> int:
        if left == right:
            return self.INF
        
        elif left == 0:
            return self.sum[right]
        
        else:
            return self.sum[right] - self.sum[left - 1]

A = [1, -1, 2, 3]
rsq = RSQPrefixSum(A)
rsq.sum, rsq.rsq(1, 3), rsq.rsq(0, 2)
```

```python
from sys import maxsize
from typing import Self
from math import log2, ceil, floor


class RMQSparseTable:
    """
    Complexity:
        - Preprocessing Time: O(N log N)
        - Query Time: O(1)
        - Update: O(N log N) (not efficient for frequent updates)
    """

    INF = maxsize

    def __init__(self: Self, data: list[int]) -> None:
        self.N = len(data)
        self.K = ceil(log2(self.N)) + 1
        self.ST = [[None] * self.N for k in range(self.K)]
        
        self.ST[0] = data.copy()
        for k in range(1, self.K):
            for i in range(self.N):
                if i + 2 ** (k - 1) < self.N:
                    self.ST[k][i] = min(
                        self.ST[k - 1][i],
                        self.ST[k - 1][i + 2 ** (k - 1)]
                    )
                else:
                    self.ST[k][i] = self.ST[k - 1][i]
        
    def rmq(self: Self, left: int, right: int) -> int:
        if left == right:
            return self.INF
        
        k = floor(log2(right - left))

        return min(
            self.ST[k][left],
            self.ST[k][right - 2 ** k]
        )

A = [0, 1, 2, 5, 8, 3, -1, 2]
rmq = RMQSparseTable(A)
rmq.ST, rmq.rmq(2, 5), rmq.rmq(0, 6)
```
