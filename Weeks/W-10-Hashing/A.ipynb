### Definitions:

1. `Injective Function`:
    For any x1 and x2, where x1 != x2 and x1, x2 belongs to X (Dm(f)), the images f(x1), f(x2), which belong to Y (Im(f)) -> f(x1) != f(x2). Therefore, different elements from the domain map to different elements in the codomain.

2. `Surjective Function`:
    For every y in Y (Im(f)), there exists at least one x in X (Dm(f)) such that f(x) = y. Therefore, every element in the codomain has at least one pre-image in the domain. Therefore, the function covers the entire codomain, there is no element in Y that is left unmapped.

3. `Bijective Function`:
    A function that is both injective and surjective. Therefore, every element in the domain maps to a unique element in the codomain, and every element in the codomain has exactly one pre-image in the domain. Therefore, there is a perfect one-to-one correspondence between elements of X and Y.

```python
from typing import Self, List

class StudentsCounter:

    def __init__(self: Self, students_names: List[str]) -> None:
        self.number_of_distinct_names = self.count_distinct_names(students_names=students_names)

    def count_distinct_names(self: Self, students_names: List[str]) -> int:
        return len({name for name in students_names})


if __name__ == "__main__":
    
    N = int(input())
    students_names = [input().strip() for _ in range(N)]

    students = StudentsCounter(students_names)
    print(students.number_of_distinct_names)
```

```python
from sympy import isprime
import random

alphabet = 'abcdefghijklmnopqrstuvwxyz'

p0 = p = len(alphabet) + 1

while not isprime(p):
    p += 1

print("Initial p:", p0)
print("Chosen prime p:", p)
a = random.randint(p0, p - 1)


s = 'kristina'
print("String:", s)
print("Length:", len(s))

h = [0] * (len(s) + 1)
a_pow = [1] * (len(s) + 1)

for i in range(len(s)):
    si = alphabet.index(s[i]) + 1
    print(f"Processing character: {s[i]} | Index in alphabet: {alphabet.index(s[i]) + 1} | h[{i}]={h[i]} | s[{i}]={si} | a_pow[{i}]={a_pow[i]} | Next Value {(h[i] + si * a_pow[i]) % p}")
    h[i + 1] = (h[i] + si * a_pow[i]) % p
    a_pow[i + 1] = (a_pow[i] * a) % p

print("Hash values:", h)
print("Powers of a:", a_pow)
```

```python
from sympy import isprime

alphabet = 'abcdefghijklmnopqrstuvwxyz'

p = len(alphabet) + 1
while not isprime(p):
    p += 1

print("Chosen prime p:", p)

s = 'kristina'
print("String:", s)
print("Length:", len(s))

h = [0] * (len(s) + 1)
a_pow = [1] * (len(s) + 1)

for i in range(len(s)):
    char_value = alphabet.index(s[i]) + 1  # Convert character to number (1-26)
    h[i + 1] = (h[i] + p * char_value) % p
    a_pow[i + 1] = (a_pow[i] * p) % p

print("Hash values:", h)
print("Powers of a:", a_pow)
```
