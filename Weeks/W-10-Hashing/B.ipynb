```python
from typing import List, Set


class DistinctSubstringCounter:

    MOD1: int = 1_000_000_007
    MOD2: int = 1_000_000_009
    BASE: int = 911382323

    def __init__(self, s: str) -> None:
        self.s: str = s
        self.n: int = len(s)

        self.h1: List[int] = [0] * (self.n + 1)
        self.h2: List[int] = [0] * (self.n + 1)
        self.p1: List[int] = [1] * (self.n + 1)
        self.p2: List[int] = [1] * (self.n + 1)

        self._build_prefix_hashes()

    def _build_prefix_hashes(self) -> None:
        for i, ch in enumerate(self.s):
            x = ord(ch)
            self.h1[i + 1] = (self.h1[i] * self.BASE + x) % self.MOD1
            self.h2[i + 1] = (self.h2[i] * self.BASE + x) % self.MOD2
            self.p1[i + 1] = (self.p1[i] * self.BASE) % self.MOD1
            self.p2[i + 1] = (self.p2[i] * self.BASE) % self.MOD2

    def _substring_hash_key(self, l: int, r: int) -> int:
        # hash1(s[l:r])
        a = (self.h1[r] - self.h1[l] * self.p1[r - l]) % self.MOD1
        # hash2(s[l:r])
        b = (self.h2[r] - self.h2[l] * self.p2[r - l]) % self.MOD2
        # Shift binary a 32 bits to the left and combine with b: (Witchcraft :D) 
        # AND bitwise pack into one int to avoid tuple overhead
        return (a << 32) | b

    def count_distinct_substrings(self) -> int:
        total = 0

        for length in range(1, self.n + 1):
            seen: Set[int] = set()
            end = self.n - length + 1

            for l in range(end):
                r = l + length
                seen.add(self._substring_hash_key(l, r))

            total += len(seen)

        return total


if __name__ == "__main__":
    # string = input().strip()
    string, raw_output = prepare_data(exercise_id="B", sample_number=3)
    counter = DistinctSubstringCounter(string)
    print(counter.count_distinct_substrings())
```
